Prompting Techniques

1. Zero- shot Prompting -> This is prompting technique where we provide all the instruction in one go 

2. Few - shot Prompting -> The prompting technique gives scope to a language model to ask something having the final result. We provide more context with few conversation.

3. Role -based Prompting -> This prompting involves assigning roles to a language model to get more accuracte result. 

Advanced Technique

4. Chain of Thought(CoT) :- Prompt our language model such that it has a scope of asking questions to a user whenever is needed.

5. Tree of Thoughts (ToT) :- LLM can have multiple response simultaneoulsy for the same query.
